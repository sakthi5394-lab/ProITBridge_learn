### 2.1 Single Object Detecion Using cv2 5.0.0

```python
import cv2

# Load image
image = cv2.imread("vijay.jpg")
grey_image = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)

# Create face detector
face_detector = cv2.FaceDetectorYN.create(
    "face_detection_yunet_2026may.onnx",
    "",
    (64, 64),
    0.6,
    0.3,
    5000
)

# Set input image size
height, width = grey_image.shape[:2]
face_detector.setInputSize((width, height))

# Detect faces
_, faces = face_detector.detect(grey_image)

# Check detection
if faces is not None:

    print("Number of faces:", len(faces))

    for face in faces:

        x = int(face[0])
        y = int(face[1])
        w = int(face[2])
        h = int(face[3])

        confidence = face[14]

        print("Face detected")
        print("X:", x)
        print("Y:", y)
        print("Width:", w)
        print("Height:", h)
        print("Confidence:", confidence)

        # Draw rectangle
        cv2.rectangle(
            grey_image,
            (x, y),
            (x + w, y + h),
            (100, 0, 0),
            2
        )

# Display result
cv2.imshow("Face Detection", grey_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### 2.3 Webcam Detection

```python
# Create face detector
face_detector = cv2.FaceDetectorYN.create(
    "face_detection_yunet_2026may.onnx",
    "",
    (64, 64),
    0.6,
    0.3,
    5000
)

video = cv2.VideoCapture(0)

while True:
    success,frame = video.read()
    if success == True:
        # Set input image size
        height, width = frame.shape[:2]
        face_detector.setInputSize((width, height))
        
        # Detect faces
        _, faces = face_detector.detect(frame)
        
        # Check detection
        if faces is not None:
        
            print("Number of faces:", len(faces))
        
            for face in faces:
        
                x = int(face[0])
                y = int(face[1])
                w = int(face[2])
                h = int(face[3])
        
                confidence = face[14]
        
                print("Face detected")
                print("X:", x)
                print("Y:", y)
                print("Width:", w)
                print("Height:", h)
                print("Confidence:", confidence)
        
                # Draw rectangle
                cv2.rectangle(
                    frame,
                    (x, y),
                    (x + w, y + h),
                    (100, 0, 0),
                    2
                )
                text = f"Confidence: {confidence:.2f}"

                cv2.putText(
                    frame,
                    text,
                    (x, y - 10),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.6,
                    (0, 255, 0),
                    2
                )
        
        # Display result
        cv2.imshow("Face Detection", frame)
        key = cv2.waitKey(10)
        if key == 81 or key == 113:
            break

video.release()
cv2.destroyAllWindows()
```
