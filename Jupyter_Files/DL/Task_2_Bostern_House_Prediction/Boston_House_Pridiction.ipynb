## 1. Import Necessary Library

```python
import numpy as np
import keras
from keras.models import Sequential
from keras.layers import Dense
from keras.datasets import boston_housing
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

## 2. Data Collection

```python
(X_train, y_train), (X_test, y_test) = boston_housing.load_data()
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

## 3. Data Preparation

```python
## Stadarditation 
```

```python
scaller = StandardScaler()
```

```python
X_train_scall = scaller.fit_transform(X_train)
X_test_scall = scaller.fit_transform(X_test)
```

```python
X_train_scall.mean().round(),X_train_scall.std().round()
```

```python
pd.DataFrame(X_train_scall)
```

```python
pd.DataFrame(X_train)
```

## 4. Module Building

```python
bos_module = Sequential()
```

```python
bos_module.add(Dense( units = 64,activation="relu",input_shape = (13,))) ##Input Layer
bos_module.add(Dense(32,activation= "relu"))
bos_module.add(Dense(10,activation= "relu"))
bos_module.add(Dense(1)) ## OuptLayer
```

```python
bos_module.compile(optimizer='adam',loss="mse",metrics = ["mae"])
```

## 5. Module Training | Module Test | Module Evaluation

```python
bos_module.fit(x=X_train_scall,
    y=y_train,
    batch_size=20,
    epochs=100,
    verbose=1,
    validation_split = 0.2)
```

```python
bos_module.summary()
```

## Module Evaluation

```python
loss,mae = bos_module.evaluate(X_test_scall,y_test)
```

```python
loss,mae
```

## Model Prediction

```python
# Example new data (13 features)
new_data = np.array([[0.02, 18, 2.5, 0, 0.45, 6.5, 65, 4.5, 1, 300, 15, 390, 5]])

# Apply same scaling
new_data = scaller.transform(new_data)

# Predict
prediction = bos_module.predict(new_data)

print("Predicted House Price:", prediction)
```
