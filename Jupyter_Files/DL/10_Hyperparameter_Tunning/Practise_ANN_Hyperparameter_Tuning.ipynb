## 1. Import Necessary Library

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

## 2. Load Data

```python
dia_data = pd.read_csv("diabetes.csv")
```

```python
X_train,X_test,y_train,y_Test = train_test_split(X,y,test_size=0.2,shuffle=True,random_state=12,stratify=y)
```

## 3, Data Preprocessing

```python
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

### Steps for the Hyperparameter Tuning

    Wrapper Class
    |
    Combine the Module Architecture
    |
    Pass the GridSearchCV to find the BestParameter
    |
    Retrain the Module with Optimal Parameter
    |
    Get the Intelligent File(.h5)"

```python
### Import the required Libarary
import numpy as np
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout
from scikeras.wrappers import KerasClassifier
from sklearn.model_selection import GridSearchCV,KFold
```

```python
def create_module(learning_Rate = 0.01,dropout_rate = 0.0,act_func = 'relu',init = 'glorot_uniform',n2 = 4,n1 = 8):
    model = Sequential()
    model.add(Dense(n1,kernel_initializer = init,activation = act_func,input_shape=(8,)))
    model.add(Dropout(dropout_rate))
    model.add(Dense(n2,kernel_initializer= init,activation=act_func))
    model.add(Dropout(dropout_rate))
    model.add(Dense(1,activation='sigmoid'))
    
    Optimizer = keras.optimizers.Adam(learning_Rate)
    model.compile(Optimizer,loss = 'binary_crossentropy',metrics = ['accuracy'])
    return model
```

```python
model = KerasClassifier(model=create_module,verbose = 1)
```

```python
param_grid = {'model__learning_Rate': [0.001, 0.01],
              #if its model_lr = Just variable naming, but here it is model__learning_rate -> Navigate inside object
              'model__dropout_rate': [0.0, 0.2],
              'model__act_func': ['relu', 'tanh'],
              'model__init': ['glorot_uniform', 'he_uniform'],
              'model__n2':[4, 8],
              'model__n1':[8, 16],
              'batch_size': [10, 20],
              'epochs': [10, 50]
            }
```

```python
K_fole_param = KFold(n_splits=3, shuffle=True, random_state=12)
grid = GridSearchCV(estimator = model,param_grid = param_grid, cv=K_fole_param,verbose=10,n_jobs = -1)
```

```python
grid_result = grid.fit(X_train,y_train)
```

```python
print("Best of %f using %s" % (grid_result.best_score_, grid_result.best_params_))
```

```python
means = grid_result.cv_results_['mean_test_score']
stds = grid_result.cv_results_['std_test_score']
params = grid_result.cv_results_['params']

for mean, std, param in zip(means, stds, params):
    print("%f (%f) with: %s" % (mean, std, param))
```

```python
pd.DataFrame(grid_result.cv_results_).sort_values(by = 'rank_test_score')
```

```python
opt_Model = create_module(learning_Rate = 0.01,dropout_rate = 0.0,act_func = 'relu',init = 'glorot_uniform',n2 = 8,n1 = 8)
#Best of 0.716563 using {'batch_size': 10, 'epochs': 50, 'model__act_func': 'relu', 'model__dropout_rate': 0.0, 'model__init': 'glorot_uniform', 'model__learning_Rate': 0.01, 'model__n1': 8, 'model__n2': 8}
```

```python
opt_Model.fit(x=X_train,
    y=y_train,
    batch_size=10,
    epochs=50,
    verbose='2')
```
