## 1. Import Necessary Library

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import keras
from keras.models import Sequential
from keras.layers import Dense
```

## 2. Daat Collection

```python
diabetes_data = pd.read_csv("diabetes.csv")
```

```python
diabetes_data
```

## 3. Data Valiodation

```python
diabetes_data.dtypes
```

```python
diabetes_data.isna().sum()
```

## 4. Data Preparartion

```python
X = diabetes_data.drop(labels = "Outcome",axis = 1)
```

```python
y= diabetes_data["Outcome"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size= 0.80,random_state = 12,shuffle= True,stratify=y)
```

```python
X_train.shape,y_train.shape
```

```python
X_test.shape,y_test.shape
```

### 5. Data PreProcessin (Standadise)

```python
scaller = StandardScaler()
```

```python
X_train = scaller.fit_transform(X_train)
```

```python
X_test = scaller.transform(X_test)
```

```python
X_train.std().round(),X_train.mean().round()
```

```python
X_test.std().round(),X_test.mean().round()
```

## 6. Module Building

```python
model = Sequential()
```

```python
model.add(Dense(8,activation = "relu",kernel_initializer='uniform',input_dim=8)) ##Input Layer
model.add(Dense(20,activation= "relu"))
model.add(Dense(10,activation= "relu"))
model.add(Dense(1,activation = "sigmoid"))
```

```python
model.compile(optimizer='Adam',loss="BinaryCrossentropy",metrics=['accuracy'])
```

## 7. Module Training

```python
model.fit(X_train,y_train,batch_size = 40,epochs=30,validation_split = 0.2)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
```

## Module testing

```python
y_Predict = model.predict(X_test)
```

```python
y_Predict = (y_Predict>0.5).astype("int")
```

```python
y_Predict
```

## Module Evalution

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
print(confusion_matrix(y_test,y_Predict))
```

```python
print(classification_report(y_test,y_Predict))
```
